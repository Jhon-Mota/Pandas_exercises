```python
import pandas as pd
df = pd.read_csv('./Automobile_data.csv')
```

```python
df.tail()
```

```python
# Exercise 2: Clean the dataset and update the CSV file
df.fillna("None")
```

```python
# Exercise 3: Find the most expensive car company name
#Print most expensive car’s company name and price.
most_expansive_car = df['price'].max()

print("Most expansive car: " + str(most_expansive_car))
```

```python
# Exercise 4: Print All Toyota Cars details
all_toyota_cars = df[df['company'] == 'toyota']
all_toyota_cars
```

```python
# Exercise 5: Count total cars per company

total_cars_per_company = df.groupby('company')['company'].value_counts().sort_values(ascending=False)
total_cars_per_company
```

```python
# Exercise 6: Find each company’s Higesht price car
higesht_price_of_each_company = df.groupby('company')['price'].max()
higesht_price_of_each_company
```

```python
#Exercise 7: Find the average mileage of each car making company

average_mileage = df.groupby('company')['average-mileage'].mean()
average_mileage
```

```python
# Exercise 8: Sort all cars by Price column
df.sort_values(by='price', ascending=True)
```

```python
# Exercise 9: Concatenate two data frames using the following conditions
GermanCars = pd.DataFrame({'Company': ['Ford', 'Mercedes', 'BMV', 'Audi'], 'Price': [23845, 171995, 135925 , 71400]})
japaneseCars = pd.DataFrame({'Company': ['Toyota', 'Honda', 'Nissan', 'Mitsubishi '], 'Price': [29995, 23600, 61500 , 58900]})

concatenated_tables = pd.concat([GermanCars, japaneseCars], keys=['German', 'Japan'])
concatenated_tables
```

```python
# Exercise 10: Merge two data frames using the following condition
# Create two data frames using the following two Dicts, Merge two data frames, and append the second data frame as a new column to the first data frame.
Car_Price = pd.DataFrame({'Company': ['Toyota', 'Honda', 'BMV', 'Audi'], 'Price': [23845, 17995, 135925 , 71400]})
car_Horsepower = pd.DataFrame({'Company': ['Toyota', 'Honda', 'BMV', 'Audi'], 'horsepower': [141, 80, 182 , 160]})

merged_tables = pd.merge(Car_Price, car_Horsepower, on='Company')
merged_tables
```
